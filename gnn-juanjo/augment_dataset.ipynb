## Augment Train and Validation Datasets

- 10 with `complexity = 45`.
- 10 with `complexity = 70`.

```python
augment_tools.augmentData(10, KNOT_DATASET_PATH, "knot_data_aug20.csv", complexity=20)
augment_tools.augmentData(10, LINK_DATASET_PATH, "link_data_aug20.csv", complexity=20)
```

```python
augment_tools.augmentData(10, KNOT_DATASET_PATH, "knot_data_aug45.csv", complexity=45)
augment_tools.augmentData(10, LINK_DATASET_PATH, "link_data_aug45.csv", complexity=45)
```

```python
augment_tools.augmentData(10, KNOT_DATASET_PATH, "knot_data_aug70.csv", complexity=70)
augment_tools.augmentData(10, LINK_DATASET_PATH, "link_data_aug70.csv", complexity=70)
```

Join dataframes into a single set

```python
sizes = ["00", "20", "45", "70"]
links_dbs = []
knots_dbs = []

for size in sizes:
    knots_db = pd.read_csv(f"datasets/knot_data_aug{size}.csv")
    knots_db = knots_db[pd.to_numeric(knots_db.Volume, errors='coerce').notnull()]
    knots_dbs.append(knots_db)

    links_db = pd.read_csv(f"datasets/link_data_aug{size}.csv")
    links_db = links_db[pd.to_numeric(links_db.Volume, errors='coerce').notnull()]
    links_dbs.append(links_db)


pd.concat(links_dbs, axis=0, ignore_index=True).to_csv("datasets/link_data.csv", index=False, float_format='%.15f')
pd.concat(knots_dbs, axis=0, ignore_index=True).to_csv("datasets/knot_data.csv", index=False, float_format='%.15f')
```
