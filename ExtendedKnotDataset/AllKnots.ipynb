```python
MyList=snappy.HTLinkExteriors(alternating=True, knots_vs_links="knots")
```

```python
print(MyList)
```

```python
MyList[0].name()[1]
```

```python
predb=[]
for L in MyList:
    predb.append([L.name(), L.num_cusps(), L.volume()])
```

```python
ak15c_database=pd.DataFrame(predb, columns=["name", "num_cusps", "volume"])
```

```python
crossing_numbers = [L.name()[1] for L in MyList]
```

```python
ak15c_database["crossing_number"] = crossing_numbers
```

```python
ak15c_database.head()
```

```python
pdcode=[]
dtcode=[]
for M in MyList:
    code=M.DT_code()
    dtcode.append(code)
    L=M.link()
    code=L.PD_code()
    pdcode.append(code)
print(pdcode[0:5])
```

```python
ak15c_database["DT_code"] = dtcode
ak15c_database["PD_code"] = pdcode
```

```python
ak15c_database.head()
```

```python
ak15c_database.to_csv("ak15c_database.csv", index=True)
```
